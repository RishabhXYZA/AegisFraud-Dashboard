**TRANSACTIONS DATASET — EDA**

```python
print("First 10 rows:")
transactions.head(10)
```

```python
# Shape
print("\nShape:")
print(transactions.shape)

# Columns
print("\nColumns:")
print(transactions.columns.tolist())

# Data types
print("\nData Types:")
print(transactions.dtypes)
```

```python
# Dataset information
print("Dataset Information:")
transactions.info()
```

```python
# Statistical summary — numerical columns
print("\nStatistical Summary — Numerical:")
transactions.describe(include=["int64", "float64"])
```

```python
# Statistical summary — categorical/object columns
print("Statistical Summary — Categorical:")
transactions.describe(include="object")
```

```python
# Missing values
print("\nNull Values:")
print(transactions.isnull().sum())

# Duplicate rows
print("\nDuplicate Rows:")
print(transactions.duplicated().sum())
```

```python
# Unique values
print("\nUnique Values:")
print(transactions.nunique())
```

```python
# Fraud distribution
print("\nFraud Distribution:")
print(transactions["Is Fraud?"].value_counts())

# Fraud percentage
print("\nFraud Percentage:")
print(
    transactions["Is Fraud?"]
    .value_counts(normalize=True)
    .mul(100)
    .round(3)
)
```

```python
# Error distribution including missing values
print("Transaction Errors — Including Missing Values:")
print(transactions["Errors?"].value_counts(dropna=False))

# Merchant State distribution including missing values
print("\nMerchant State Distribution — Including Missing Values:")
print(transactions["Merchant State"].value_counts(dropna=False))

# Use Chip distribution
print("\nUse Chip Distribution:")
print(transactions["Use Chip"].value_counts())

# Top Merchant Names
print("\nTop 10 Merchants:")
print(transactions["Merchant Name"].value_counts().head(10))

# Top Merchant Cities
print("\nTop 10 Merchant Cities:")
print(transactions["Merchant City"].value_counts().head(10))

# MCC distribution
print("\nTop 10 MCC Values:")
print(transactions["MCC"].value_counts().head(10))
```
